```python
import pandas as pd
import plotly.express as px
```

```python
url = '../data/clean/Cleaned Employment, average hourly and weekly earnings (including overtime),.csv'
df = pd.read_csv(url)
df = df[(df['GEO']!= 'Yukon') & (df['GEO']!='Northwest Territories') & (df['GEO']!='Nunavut')]
df
```

```python
for _ in df['Estimate'].value_counts().index[0:5]:
    fig = px.line(df[(df['Estimate'] == _)],x = 'REF_DATE', y = 'VALUE', color = 'GEO', title=_)
    fig.show()
```

```python
for _ in df['Estimate'].value_counts().index[0:-3]:
    fig = px.line(df[(df['Estimate'] == _) & (df['GEO'] == 'Nova Scotia')],x = 'REF_DATE', y = 'VALUE', color = 'GEO', title=_)
    fig.show()
```
