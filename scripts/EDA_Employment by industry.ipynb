```python
import pandas as pd
import plotly.express as px
```

```python
url = '../data/clean/Cleaned Employment by industry.csv'
df = pd.read_csv(url)
df = df[df['Data type'] == 'Seasonally adjusted']
df = df[df['Data type'] == 'Seasonally adjusted']
df
```

```python
for province in df['GEO'].value_counts().index:
    fig = px.line(df[(df['GEO'] == province) & ((df['North American Industry Classification System'] != 'Total employed, all industries' ))],
                  x='REF_DATE',
                  y='VALUE', 
                  color = ('North American Industry Classification System'), 
                  title= f'{province} Workforce Makeup')
    fig.show()
```

```python
fig = px.line(df[(df['GEO'] != 'Canada') & (df['North American Industry Classification System'] == 'Services-producing sector')],
              x='REF_DATE',
              y='VALUE', 
              color = 'GEO', 
              title= f'Province Services Workforce Makeup')
fig.show()
```

```python
for province in df['GEO'].value_counts().index:
    fig = px.line(df[(df['GEO'] == province) & (df['North American Industry Classification System'] == 'Services-producing sector')],
                  x='REF_DATE',y='VALUE', 
                  title= f'{province} Services Workforce Makeup')
    fig.show()
```

```python
for province in df['GEO'].value_counts().index:
    fig = px.line(df[(df['GEO'] == province) & (df['North American Industry Classification System'] == 'Professional, scientific and technical services')],
                  x='REF_DATE',y='VALUE', 
                  title= f'{province} Professional Workforce Makeup')
    fig.show()
```
