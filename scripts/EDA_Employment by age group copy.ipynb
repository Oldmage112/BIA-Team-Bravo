```python
import pandas as pd
import plotly.express as px
```

```python
url = '../data/clean/Cleaned Labour force by educational attainment.csv'
df = pd.read_csv(url, index_col=0)
df
```

```python
df2 = df[(df['Age group'] == '25 to 54 years') 
   ]
df2
```

```python
for edu in df2['Educational attainment'].value_counts().index:
    fig = px.line(df2[(df2['Educational attainment'] == edu)], 'REF_DATE', 'VALUE', color = 'GEO', title=f'{edu} Employment by Province')
    fig.show()
```

```python
df3 = df[(df['Gender'] == 'Total - Gender')
   & ((df['Labour force characteristics'] == 'Full-time employment') | (df['Labour force characteristics'] == 'Part-time employment'))
   & (df['Age group'] == '15 years and over')
   & (df['Data type'] == 'Seasonally adjusted')
]
df3
```

```python
for province in df['GEO'].value_counts().index:
    fig = px.line(df3[(df['GEO'] == province)], 'REF_DATE', 'VALUE', color = 'Labour force characteristics', title=f'{province} Employment by Part-Time vs. Full-Time')
    fig.show()
```

```python
df4 = df[(df['Labour force characteristics'] == 'Participation rate') 
   & (df['Gender'] == 'Total - Gender')
   & (df['Data type'] == 'Seasonally adjusted')
   ]
```

```python
for province in df['GEO'].value_counts().index:
    fig = px.line(df4[(df['GEO'] == province)], 'REF_DATE', 'VALUE', color = 'Age group', title=f'{province} Employment by Part-Time vs. Full-Time')
    fig.show()
```
