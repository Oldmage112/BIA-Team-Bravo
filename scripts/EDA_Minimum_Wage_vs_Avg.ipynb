```python
import pandas as pd
import plotly.express as px
```

```python
url = '../data/clean/NS minimum wage.csv'
df = pd.read_csv(url, index_col=0)
df = df[df['Experience Level'] != 'Experienced']
```

```python
df
```

```python
px.line(df,x = 'Effective Date', y = 'Minimum Wage Rate')
```

```python
url2 = '../data/clean/Cleaned Employment, average hourly and weekly earnings (including overtime),.csv'
df2 = pd.read_csv(url2)
df2 = df2[(df2['GEO']== 'Nova Scotia')]
df2
```

```python
fig = px.line(
    df,
    x="Effective Date",
    y="Minimum Wage Rate",
    title="Minimum Wage vs Estimates",
    labels='Minimum Wage'
)

for estimate in df2["Estimate"].value_counts().index:
    df3 = df2[(df2["Estimate"] == 'Average hourly earnings including overtime for hourly employees') | 
       (df2['Estimate'] == 'Average hourly earnings including overtime for salaried employees')]
    subset = df3[df3["Estimate"] == estimate]

    fig.add_scatter(
        x=subset["REF_DATE"],
        y=subset["VALUE"],
        mode="lines",
        name=str(estimate)
    )

fig.show()
```
